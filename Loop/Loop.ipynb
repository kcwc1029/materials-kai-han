# 迴圈

## for迴圈

- [練習：for迴圈練習題目](./HW/for迴圈練習題目.ipynb)

for 迴圈適合用在已知要重複執行幾次，或需要逐一讀取資料的情況。Python 會依序取出指定範圍或集合中的元素，每取出一個元素，就執行一次迴圈內的程式碼，直到所有元素都處理完畢。

![](./assets/for迴圈.png)

### 範例：for迴圈重複印出內

````python
for i in range(5):
	print("哈哈")
````

### 範例：for迴圈重複印出i

````python
for i in range(5):
	print(i)
````

### range()：產生一個整數序列

range() 函式是一種內建的可迭代物件，主要用於產生一個整數序列，通常用於 for 迴圈中以控制迴圈執行的次數。

```py
range(start, stop, step)
# 產生從 start 開始，
# 到 stop-1 結束，
# 每次以 step 為步長的整數序列。step 可以是正數（遞增）或負數（遞減）。
```

````python
for i in range(5):
    print(i)

for i in range(1, 5):
    print(i)

for i in range(1, 10, 2):
    print(i)

for i in range(10, 0, -2):
    print(i)
````

### 範例：倒數計時器

````python
print("跨年倒數開始：")

# range(5, 0, -1) 的意思是：從 5 開始，跑到 0 為止(不包含0)，每次遞減 1
for i in range(5, 0, -1):
    print(i)
    
print("新年快樂！")
````

### 範例：for迴圈用於陣列點名

把一個列表（List）裡面的東西，依序拿出來處理。

````python
students = ["小明", "小華", "小美", "阿強"]

print("開始點名：")
# "student" 只是一個暫時的代名詞，代表每次拿出來的那個人
for student in students:
    print(f"{student} 到了！")
    
print("點名結束。")
````

### 範例：撲滿存錢筒

````python
savings = [50, 100, 30, 20, 150, 0, 200]

# 準備一個變數當作撲滿，初始值為 0
total_money = 0

for money in savings:
    # 每次迴圈執行時，把當天的錢加到撲滿裡
    total_money = total_money + money  # 也可以簡寫成 total_money += money
    print(f"今天存了 {money} 元，撲滿裡有 {total_money} 元了。")

print(f"這禮拜總共存了：{total_money} 元！")
````

## 雙重for迴圈

### 範例：雙重for迴圈用於疊代兩個陣列

````python
colors = ["Red", "Green", "Blue"]
shapes = ["Circle", "Square", "Line"]

for color in colors:
    for shape in shapes:
        print(f"顏色: {color}, 形狀: {shape}")
````

### 範例：雙重for迴圈 九九乘法表

````python
for i in range(1, 10): # 控制被乘數，從 1 到 9
    for j in range(1, 10): # 控制乘數，從 1 到 9
        print(f"{i} x {j} = {i * j}", end="\t") # 印出乘法結果，使用 Tab 分隔
    print() # 每完成一輪內層迴圈就換行
````

### break：強制離開 for 迴圈

````python
scores = [85, 92, 78, 95, 88]
for score in scores:
    if score > 90:
        print(f"找到第一個超過 90 分的成績：{score}")
        break # 找到後立即終止迴圈
````

### continue：for迴圈暫停但不終止

````python
### 範例：只列印偶數
for i in range(10):
    if i % 2 != 0:
        continue # 如果是奇數，跳過本次迴圈
    print(i)
````

## while迴圈

while 迴圈適合用在不確定需要重複執行幾次，但知道停止條件的情況。每次執行迴圈之前，Python 都會檢查條件是否成立。如果條件為 True，就繼續執行；如果條件為 False，就結束迴圈。使用時要特別注意更新條件相關的變數，避免產生無窮迴圈。

![](./assets/while迴圈.png)

### 範例：使用者輸入q時終止

````python
msg = ''
while msg != 'q':
    msg = input("請輸入訊息（輸入 'q' 結束）：")
    print(msg)
````

### 範例：計算使用者輸入數字的總和，輸入 0 終止

````python
total = 0
num = int(input("請輸入一個數字（輸入 0 結束）："))

while num != 0:
    total += num
    num = int(input("請輸入一個數字（輸入 0 結束）："))

print(f"總和為: {total}")
````

### 範例：雙重while迴圈 九九乘法表

````python
i = 1 # 設定外層迴圈初始值

while i <= 9: # 控制被乘數，從 1 到 9
    j = 1 # 每次進入外層迴圈時，重設乘數為 1

    while j <= 9: # 控制乘數，從 1 到 9
        print(f"{i} x {j} = {i * j}", end="\t") # 印出乘法結果，使用 Tab 分隔
        j += 1 # 乘數增加 1

    print() # 每完成一輪內層迴圈就換行
    i += 1 # 被乘數增加 1
````
