## Python 邏輯運算子

邏輯運算子通常會用在「判斷條件」裡面。

例如：

你想判斷一個學生：

- 分數是否及格
- 是否同時符合兩個條件
- 是否只要符合其中一個條件就可以
- 是否不是某種狀態

這時候就會用到邏輯運算子。

| 運算子 | 意思 | 說明                   |
| ------ | ---- | ---------------------- |
| `and`  | 而且 | 兩個條件都要成立       |
| `or`   | 或者 | 其中一個條件成立就可以 |
| `not`  | 不是 | 將結果反轉             |

````python
### 範例：兩個條件都要成立
score = 85
attendance = 90
# 兩個條件都要成立：AND
result = score >= 60 and attendance >= 80

print(result)
````

````python
### 範例：是否可以通過課
score = 75
attendance = 60

can_pass = score >= 60 and attendance >= 80

print(can_pass)
````

````python
### 範例：其中一個條件成立就可以
has_student_id = True
has_qrcode = False

can_enter = has_student_id or has_qrcode

print(can_enter)
````

````python
### 範例：是否可以享有優惠
# 是否學生
is_student = False
# 是否是會員
has_coupon = True

can_discount = is_student or has_coupon

print(can_discount)
````

````python
### 範例：把結果反過來
is_raining = False

print(not is_raining)
````

````python
### 範例：登入判斷
username = "admin"
password = "1234"

is_login = username == "admin" and password == "1234"

print(is_login)
````

## 判斷式(if / else)

在寫程式時，我們常常會遇到一種情況：根據不同條件，執行不同動作。

例如：

- 成績及格才能通過
- 年齡滿 18 才能註冊
- 餘額不足不能買東西
- 下雨就帶傘

這時候就會用到「判斷式」。

### if：如果條件成立，就執行

```py
score = 80

if score >= 60:
    print("及格")
```

````python
# 範例：年齡判斷
age = 20

if age >= 18:
    print("已成年")
````

### if else：二選一判斷

有時候我們希望：

- 條件成立做一件事
- 條件不成立做另一件事

````python
### 範例：餘額判斷
money = 30
drink_price = 35

if money >= drink_price:
    print("可以購買飲料")
else:
    print("錢不夠")
````


### if elif else：多條件判斷

如果有很多種情況。

例如：

- 90 分以上：A
- 80 分以上：B
- 60 分以上：C
- 其他：不及格

```py
score = 85

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 60:
    print("C")
else:
    print("不及格")
```
